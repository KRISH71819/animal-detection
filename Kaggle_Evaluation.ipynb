# Animal Guard — Full Model Evaluation (10-Class Indian Farm Animal Detection)

**Model:** YOLO11n (100 epochs, `best.pt`) | **GPU:** NVIDIA T4 | **Image Size:** 640 | **Batch:** 32  
**Classes:** 10 (cow, buffalo, goat, dog, wild_boar, monkey, nilgai, elephant, deer, bird)  
**Final Training Metrics (Epoch 100):** mAP@0.5 = **84.0%** | mAP@0.5:0.95 = **61.5%**

**Instructions:** Attach the `kaggle-eval-pack` dataset then click **Run All**

```python
# CELL 1: Setup working directory & copy files
import os, shutil, sys
from pathlib import Path

INPUT_ROOT   = '/kaggle/input'
WORKING_DIR  = '/kaggle/working/animal-guard'
WEIGHTS_PATH = os.path.join(WORKING_DIR, 'runs', 'animal_guard_train', 'weights', 'best.pt')

SRC_DIR = None
for root, dirs, files in os.walk(INPUT_ROOT):
    if 'evaluate.py' in files and 'config.py' in files:
        SRC_DIR = root
        break

if SRC_DIR is None:
    raise FileNotFoundError('evaluate.py and config.py not found under /kaggle/input')

print(f'[OK] Source found: {SRC_DIR}')

if not os.path.exists(WORKING_DIR):
    shutil.copytree(SRC_DIR, WORKING_DIR)
    print(f'[OK] Files copied to {WORKING_DIR}')
else:
    for f in ['config.py', 'evaluate.py', 'collect_data.py', 'requirements.txt', 'verified_datasets.json']:
        src = os.path.join(SRC_DIR, f)
        if os.path.exists(src):
            shutil.copy2(src, os.path.join(WORKING_DIR, f))
    print('[OK] Scripts refreshed')

os.chdir(WORKING_DIR)
sys.path.insert(0, WORKING_DIR)

# Locate best.pt (try multiple paths, fall back to last.pt)
CANDIDATES = [
    WEIGHTS_PATH,
    os.path.join(WORKING_DIR, 'best.pt'),
    os.path.join(SRC_DIR, 'best.pt'),
    os.path.join(SRC_DIR, 'runs', 'animal_guard_train', 'weights', 'best.pt'),
    os.path.join(WORKING_DIR, 'last.pt'),
    os.path.join(SRC_DIR, 'last.pt'),
    os.path.join(SRC_DIR, 'runs', 'animal_guard_train', 'weights', 'last.pt'),
]

FOUND = next((c for c in CANDIDATES if os.path.exists(c)), None)
if FOUND is None:
    raise FileNotFoundError('No weights file found! Attach kaggle_eval_pack with best.pt')

if FOUND != WEIGHTS_PATH:
    os.makedirs(os.path.dirname(WEIGHTS_PATH), exist_ok=True)
    shutil.copy2(FOUND, WEIGHTS_PATH)
    print(f'[OK] Copied {os.path.basename(FOUND)} to {WEIGHTS_PATH}')

mb = os.path.getsize(WEIGHTS_PATH) / 1_048_576
print(f'[OK] Weights ready: best.pt ({mb:.1f} MB)')
```

```python
# CELL 2: Install dependencies
import subprocess, sys, os
os.chdir(WORKING_DIR)
result = subprocess.run(
    [sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'],
    capture_output=False
)
print('[OK] Dependencies installed' if result.returncode == 0 else '[ERROR] pip install failed')
```

```python
# CELL 3: Fresh dataset download & merge
import os, shutil, subprocess, sys
os.chdir(WORKING_DIR)

for d in [os.path.join(WORKING_DIR, 'data', 'merged'), os.path.join(WORKING_DIR, 'data', 'raw')]:
    if os.path.exists(d):
        shutil.rmtree(d)
        print(f'[i] Cleared {os.path.basename(d)}/')

print('[i] Downloading ALL datasets (~10-20 min)...')
result = subprocess.run([sys.executable, 'collect_data.py'], capture_output=False)

if result.returncode != 0:
    print('[ERROR] collect_data.py failed!')
else:
    vi = os.path.join(WORKING_DIR, 'data', 'merged', 'val', 'images')
    ti = os.path.join(WORKING_DIR, 'data', 'merged', 'train', 'images')
    nv = len(os.listdir(vi)) if os.path.exists(vi) else 0
    nt = len(os.listdir(ti)) if os.path.exists(ti) else 0
    print(f'[OK] Dataset ready! Train={nt:,}  Val={nv:,}  Total={nt+nv:,}')
```

```python
# CELL 4: GPU Validation — Full Metrics
import os, sys, json, csv
from ultralytics import YOLO

os.chdir(WORKING_DIR)
WEIGHTS_PATH = os.path.join(WORKING_DIR, 'runs', 'animal_guard_train', 'weights', 'best.pt')
DATA_YAML    = os.path.join(WORKING_DIR, 'data', 'data.yaml')

vi = os.path.join(WORKING_DIR, 'data', 'merged', 'val', 'images')
vl = os.path.join(WORKING_DIR, 'data', 'merged', 'val', 'labels')
n_val = len(os.listdir(vi)) if os.path.exists(vi) else 0
n_inst = 0
if os.path.exists(vl):
    for lf in os.listdir(vl):
        if lf.endswith('.txt'):
            with open(os.path.join(vl, lf)) as f:
                n_inst += sum(1 for ln in f if ln.strip())

print(f'[i] Val images: {n_val:,}  |  Instances: {n_inst:,}')
print('[i] Loading model...')
model = YOLO(WEIGHTS_PATH)

print(f'[i] Running validation on GPU (imgsz=640, batch=32)...')
results = model.val(data=DATA_YAML, imgsz=640, batch=32, device=0, verbose=True, plots=True)

box = results.box
CLASSES = ['cow', 'buffalo', 'goat', 'dog', 'wild_boar', 'monkey', 'nilgai', 'elephant', 'deer', 'bird']

rows = []
for i, cls in enumerate(CLASSES):
    if i < len(box.ap50):
        p  = float(box.p[i]) if i < len(box.p) else 0.0
        r  = float(box.r[i]) if i < len(box.r) else 0.0
        f1 = 2*p*r/(p+r) if (p+r) > 0 else 0.0
        rows.append({
            'Class': cls.replace('_',' ').title(),
            'Precision (%)':    round(p*100, 1),
            'Recall (%)':       round(r*100, 1),
            'F1-Score (%)':     round(f1*100, 1),
            'mAP@0.5 (%)':      round(float(box.ap50[i])*100, 1),
            'mAP@0.5:0.95 (%)': round(float(box.ap[i])*100, 1)
        })

print('\n' + '='*90)
print('  ANIMAL GUARD YOLO11n — FULL VALIDATION (100 Epochs Trained)')
print(f'  Val Images: {n_val:,}  |  Instances: {n_inst:,}')
print('='*90)
print(f"  {'Class':<15} {'Precision':>11} {'Recall':>10} {'F1-Score':>10} {'mAP@0.5':>10} {'mAP@.5:.95':>12}")
print('  ' + '-'*86)
for row in rows:
    flag = '  WARNING' if row['mAP@0.5 (%)'] < 50 else ('  low' if row['mAP@0.5 (%)'] < 70 else '')
    print(f"  {row['Class']:<15} {row['Precision (%)']:>10.1f}% {row['Recall (%)']:>9.1f}% {row['F1-Score (%)']:>9.1f}% {row['mAP@0.5 (%)']:>9.1f}% {row['mAP@0.5:0.95 (%)']:>11.1f}%{flag}")
print('  ' + '-'*86)
print(f"  {'OVERALL MEAN':<15} {box.mp*100:>10.1f}% {box.mr*100:>9.1f}% {'':>9} {box.map50*100:>9.1f}% {box.map*100:>11.1f}%")
print('='*90)

out = '/kaggle/working'
csv_p  = os.path.join(out, 'eval_results_FULL.csv')
txt_p  = os.path.join(out, 'eval_results_FULL.txt')
json_p = os.path.join(out, 'eval_results_FULL.json')

with open(csv_p, 'w', newline='', encoding='utf-8') as f:
    w = csv.DictWriter(f, fieldnames=rows[0].keys())
    w.writeheader(); w.writerows(rows)

with open(txt_p, 'w', encoding='utf-8') as f:
    f.write(f'ANIMAL GUARD YOLO11n  — 100 Epochs\n')
    f.write(f'Val Images: {n_val:,}  Instances: {n_inst:,}\n\n')
    f.write(f"{'Class':<15} {'Prec%':>6} {'Rec%':>6} {'F1%':>6} {'mAP50%':>7} {'mAP5095%':>9}\n")
    f.write('-'*60 + '\n')
    for row in rows:
        f.write(f"{row['Class']:<15} {row['Precision (%)']:>6.1f} {row['Recall (%)']:>6.1f} {row['F1-Score (%)']:>6.1f} {row['mAP@0.5 (%)']:>7.1f} {row['mAP@0.5:0.95 (%)']:>9.1f}\n")
    f.write('-'*60 + '\n')
    f.write(f"{'OVERALL':<15} {box.mp*100:>6.1f} {box.mr*100:>6.1f} {'--':>6} {box.map50*100:>7.1f} {box.map*100:>9.1f}\n")

with open(json_p, 'w') as f:
    json.dump({
        'model':'YOLO11n','epochs_trained':100,
        'val_images':n_val,'instances':n_inst,
        'map50':round(float(box.map50),4),'map50_95':round(float(box.map),4),
        'precision':round(float(box.mp),4),'recall':round(float(box.mr),4),
        'per_class':{row['Class'].lower().replace(' ','_'):{
            'precision':row['Precision (%)']/100,'recall':row['Recall (%)']/100,
            'f1':row['F1-Score (%)']/100,'ap50':row['mAP@0.5 (%)']/100,'ap50_95':row['mAP@0.5:0.95 (%)']/100
        } for row in rows}
    }, f, indent=2)

print(f'\n[OK] CSV  : {csv_p}')
print(f'[OK] TXT  : {txt_p}')
print(f'[OK] JSON : {json_p}')
print('[!] Download files from Kaggle Output tab!')
```
