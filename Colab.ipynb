```python
# ── CELL 1: Setup — Prepare files & checkpoint ──────────────────────
import os, shutil

INPUT_ROOT  = '/kaggle/input'
WORKING_DIR = '/kaggle/working/animal-guard'
WEIGHTS_DIR = WORKING_DIR + '/runs/animal_guard_train/weights'
LAST_CKPT   = WEIGHTS_DIR + '/last.pt'

# Find the uploaded dataset folder (contains train.py + config.py)
SRC_DIR = None
for root, dirs, files in os.walk(INPUT_ROOT):
    if 'train.py' in files and 'config.py' in files:
        SRC_DIR = root
        break
if SRC_DIR is None:
    raise FileNotFoundError('train.py not found under /kaggle/input — did you attach the dataset?')

SCRIPTS = [
    'config.py', 'train.py', 'evaluate.py', 'detect.py',
    'collect_data.py', 'augment_nightvision.py',
    'augment_weak_classes.py', 'hard_negative_mining.py', 'requirements.txt',
]

if not os.path.exists(WORKING_DIR):
    shutil.copytree(SRC_DIR, WORKING_DIR)
    print('[✓] Files copied from', SRC_DIR)
else:
    for f in SCRIPTS:
        src = os.path.join(SRC_DIR, f)
        dst = os.path.join(WORKING_DIR, f)
        if os.path.exists(src):
            shutil.copy2(src, dst)
    print('[✓] Scripts refreshed from', SRC_DIR)

os.chdir(WORKING_DIR)
print('[✓] Working dir:', WORKING_DIR)

# Ensure weights directory exists
os.makedirs(WEIGHTS_DIR, exist_ok=True)

# Sync checkpoint from root or runs/ if present in uploaded files
root_ckpt = os.path.join(WORKING_DIR, 'last.pt')
if os.path.exists(root_ckpt) and not os.path.exists(LAST_CKPT):
    shutil.copy2(root_ckpt, LAST_CKPT)
    print('[✓] Synced last.pt to weights directory:', LAST_CKPT)
elif os.path.exists(root_ckpt) and os.path.exists(LAST_CKPT):
    # If root_ckpt is newer or present, copy it
    shutil.copy2(root_ckpt, LAST_CKPT)
    print('[✓] Synced latest last.pt to weights directory:', LAST_CKPT)

# Copy results.csv if present in uploaded files
root_results = os.path.join(WORKING_DIR, 'results.csv')
target_results = os.path.join(WORKING_DIR, 'runs/animal_guard_train/results.csv')
if os.path.exists(root_results) and not os.path.exists(target_results):
    shutil.copy2(root_results, target_results)

if os.path.exists(LAST_CKPT):
    mb = os.path.getsize(LAST_CKPT) / 1_048_576
    print(f'[✓] Checkpoint ready: last.pt ({mb:.1f} MB) — will RESUME from Epoch 40')
else:
    print('[!] No checkpoint found — will start fresh')
```

```python
# ── CELL 2: Download & merge datasets ───────────────────────────────
import os, subprocess, sys
os.chdir(WORKING_DIR)
print('[i] Downloading and merging datasets...')
subprocess.run([sys.executable, 'collect_data.py'], check=True)
```

```python
# ── CELL 3: Night-vision IR augmentation ────────────────────────────
import os, subprocess, sys, shutil
os.chdir(WORKING_DIR)

marker = os.path.join(WORKING_DIR, 'data', 'merged', '.ir_augmentation_done')
if os.path.exists(marker):
    print('[OK] Night-vision IR augmentation already done. Skipping.')
else:
    print('[i] Running night-vision IR augmentation (ratio=0.20, JPEG saves)...')
    subprocess.run([sys.executable, 'augment_nightvision.py', '--ratio', '0.20'], check=True)
    open(marker, 'w').close()
```

```python
# ── CELL 4: Phase 6 Weak-class augmentation (Buffalo 8x, Cow 4x) ────
import os, subprocess, sys, shutil
os.chdir(WORKING_DIR)

train_img_dir = os.path.join(WORKING_DIR, 'data', 'merged', 'train', 'images')
n_before = len(os.listdir(train_img_dir)) if os.path.exists(train_img_dir) else 0

disk = shutil.disk_usage('/kaggle/working')
free_gb = disk.free / (1024 ** 3)
print(f'[i] Disk free before augmentation: {free_gb:.1f} GB')
if free_gb < 6.0:
    print(f'[!] Only {free_gb:.1f} GB free — skipping augmentation to prevent disk-full crash.')
else:
    marker = os.path.join(WORKING_DIR, 'data', 'merged', '.augmentation_done')
    if os.path.exists(marker):
        print(f'[OK] Augmentation already done ({n_before} images). Skipping.')
    else:
        print('[i] Running Phase 6 weak-class augmentation: buffalo 8x, cow 4x...')
        subprocess.run([
            sys.executable, 'augment_weak_classes.py',
            '--classes', 'buffalo', 'cow',
            '--multiplier', '8', '4',
        ], check=True)
        open(marker, 'w').close()
        n_after = len(os.listdir(train_img_dir))
        print(f'[OK] Augmentation complete: {n_before} -> {n_after} training images.')
```

```python
# ── CELL 5: Model Training — Resume from Epoch 40 ───────────────────
import os, sys, importlib, argparse
os.chdir(WORKING_DIR)
sys.path.insert(0, WORKING_DIR)

for mod_name in ['config', 'train']:
    if mod_name in sys.modules:
        importlib.reload(sys.modules[mod_name])
from train import train as run_training

LAST_CKPT = WORKING_DIR + '/runs/animal_guard_train/weights/last.pt'
is_resume = os.path.exists(LAST_CKPT)

args = argparse.Namespace(
    batch=None, epochs=None, model=None, workers=None, imgsz=None, device=None, resume=is_resume,
)

if is_resume:
    mb = os.path.getsize(LAST_CKPT) / 1_048_576
    print(f'[OK] Checkpoint found: last.pt ({mb:.1f} MB) — will RESUME from Epoch 40')
else:
    print('[OK] Starting fresh training')

run_training(args)
```

```python
# ── CELL 6: Backup weights ──────────────────────────────────────────
import os, glob, zipfile

weights_dir = WORKING_DIR + '/runs/animal_guard_train/weights'
backup_path = '/kaggle/working/weights_backup.zip'

if not os.path.exists(weights_dir):
    print('[✗] Weights directory not found — training may not have run')
else:
    pt_files = sorted(glob.glob(weights_dir + '/*.pt'))
    if not pt_files:
        print('[✗] No .pt files found — no epoch completed')
    else:
        with zipfile.ZipFile(backup_path, 'w', zipfile.ZIP_DEFLATED) as zf:
            for f in pt_files:
                zf.write(f, os.path.basename(f))
        size_mb = os.path.getsize(backup_path) / 1_048_576
        print(f'[✓] weights_backup.zip ({size_mb:.1f} MB) ready in Output tab!')
```
